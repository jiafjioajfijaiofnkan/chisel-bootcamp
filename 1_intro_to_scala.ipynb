# 理解 Scala
Scala 是另一种支持常见编程范式的编程语言。我们选择使用它有以下几个原因：

- 它是一种适合托管嵌入式 DSL 的语言。
- 它拥有一个强大而优雅的库，用于操作各种数据集合。
- 它具有严格的类型系统，有助于在开发周期的早期（即编译时）捕获大量错误。
- 它具有强大的表达和传递函数的方式。
- Chi*p*el、Chi*j*el 和 Chi*c*el 不如 Chisel 那么顺口。

当我们稍后讨论 Chisel 时，所有这些要点都会变得显而易见，但现在，我们将专注于阅读和编写 Scala 代码的基础知识。

# 条件语句

Scala 实现条件语句的方式与其他编程语言类似。

````scala
// 一个简单的条件语句；顺便说一下，这是一条注释
if (numberOfKittens > kittensPerHouse) { 
    println("小猫太多了！！！") 
}
// 当所有分支都是单行时，大括号不是必需的。但是，
// Scala 风格指南建议仅在包含“else”子句时才省略大括号。
// （最好不要这样，即使它能编译……）
if (numberOfKittens > kittensPerHouse) 
    println("小猫太多了！！！")

// if 当然有 else 子句
// 这里可以省略大括号！
if (done) 
    println("我们完成了")
else 
    numberOfKittens += 1

// 还有 else if
// 出于风格考虑，保留大括号，因为并非所有分支都是单行。
if (done) {
    println("我们完成了")
}
else if (numberOfKittens < kittensPerHouse) {
    println("更多小猫！")
    numberOfKittens += 1
}
else {
    done = true
}
````

但是在 Scala 中，“`if`”条件语句会返回一个值。这个值是什么？它由所选分支的最后一行给出。这非常强大，尤其是在用于初始化函数和类中的值时。它看起来像这样：

````scala
val likelyCharactersSet = if (alphabet.length == 26)
    "english"
else 
    "not english"

println(likelyCharactersSet)
````

我们创建了一个常量 *likelyCharactersSet*，其值在运行时有条件地确定。

# 方法 (函数)

方法使用关键字 `def` 定义。在这里，我们将滥用符号，也称它们为函数。函数参数（或形参）在逗号分隔的列表中指定，该列表指定参数的名称、类型以及可选的默认值。为了清晰起见，应指定返回类型。

不带任何参数的 Scala 函数不需要空括号。这通常会使开发人员在类成员由于引用时存在某些计算而成为函数的情况下更容易。按照惯例，没有副作用（即调用它们不会改变任何东西，它们只是返回一个值）的无参数函数不使用括号，而有副作用（可能会更改类变量或打印内容）的函数应需要括号。

## 简单声明

````scala
// 简单的缩放函数，带有一个输入参数，例如 times2(3) 返回 6
// 对于简短的单行函数，可以省略花括号。
def times2(x: Int): Int = 2 * x

// 更复杂的函数
def distance(x: Int, y: Int, returnPositive: Boolean): Int = {
    val xy = x * y
    if (returnPositive) xy.abs else -xy.abs
}
````

## 重载函数
同一个函数名可以用多种方式使用。参数及其类型决定了一个签名，编译器可以根据该签名确定应调用哪个版本的函数。不过，应避免重载函数。

````scala
// 重载函数
def times2(x: Int): Int = 2 * x
def times2(x: String): Int = 2 * x.toInt

times2(5)
times2("7")
````

## 递归和嵌套函数
花括号定义代码作用域。在函数的作用域内可以存在更多函数或递归函数调用。在特定作用域中定义的函数只能在该作用域内访问。

````scala
/** 打印一个由“X”组成的三角形
  * 这是另一种注释风格
  */
def asciiTriangle(rows: Int) {
    
    // 这很巧妙：乘以“X”会生成一个包含多个“X”副本的字符串
    def printRow(columns: Int): Unit = println("X" * columns)
    
    if(rows > 0) {
        printRow(rows)
        asciiTriangle(rows - 1) // 这是递归调用
    }
}

// printRow(1) // 这不起作用，因为我们在其作用域之外调用 printRow
asciiTriangle(6)
````

# 列表
Scala 实现了各种聚合或序列对象。列表与数组非常相似，但支持额外的附加和提取操作。

````scala
val x = 7
val y = 14
val list1 = List(1, 2, 3)
val list2 = x :: y :: y :: Nil       // 组装列表的另一种表示法

val list3 = list1 ++ list2           // 将第二个列表附加到第一个列表
val m = list2.length
val s = list2.size

val headOfList = list1.head          // 获取列表的第一个元素
val restOfList = list1.tail          // 获取一个移除了第一个元素的新列表

val third = list1(2)                 // 获取列表的第三个元素（从 0 开始索引）
````

# `for` 语句

Scala 有一个 `for` 语句，它的工作方式类似于传统的 for 语句。您
可以迭代一个值范围。

````scala
for (i <- 0 to 7) { print(i + " ") }
println()
````

使用 `until` 代替 `to` 来从 0 迭代到 6（不包括 7）。

````scala
for (i <- 0 until 7) { print(i + " ") }
println()
````

添加一个 `by` 以按某个固定量递增。以下代码打印出 0 到 10 之间的偶数。

````scala
for(i <- 0 to 10 by 2) { print(i + " ") }
println()
````

如果您有某种集合并且想要访问所有元素，您可以使用 `for`
作为*迭代器*，就像在 Java 和 Python 中一样。这里我们创建一个包含 4 个随机整数的列表，然后对它们求和。

````scala
val randomList = List(scala.util.Random.nextInt(), scala.util.Random.nextInt(), scala.util.Random.nextInt(), scala.util.Random.nextInt())
var listSum = 0
for (value <- randomList) {
  listSum += value
}
println("总和是 " + listSum)
````

Scala 的 `for` 还有很多其他技巧。它可以直观地用于
各种传统的迭代需求，但它可能不是最方便使用的。像对数组元素求和这样的操作通常更容易使用称为*推导式*的一系列丰富的函数来完成，这些函数可用于许多不同的元素集合。后续模块将提供有关 `for` 及其相关内容的更多信息。

# Scala 是一种面向对象的语言
Scala 是面向对象的，了解这一点对于充分利用 Scala 和 Chisel 非常重要。请注意，毫无疑问，描述这一切的方法不止一种。
1. 变量是对象。
1. 从 *Scala* 的 `val` 声明的意义上说，常量也是对象。
1. 即使是字面值也是对象。
1. 即使是函数本身也是对象。稍后会详细介绍。
1. 对象是类的实例。
  1. 事实上，在 Scala 中几乎所有重要方面，*面向对象*中的*对象*都将被称为实例。
1. 在定义类时，程序员指定：
  1. 与类关联的数据（`val`、`var`）。
  1. 类的实例可以执行的操作，称为方法或函数。
1. 类可以扩展其他类。
  1. 被扩展的类是超类；扩展者是子类。
  1. 在这种情况下，子类从超类继承数据和方法。
  1. 类可以通过许多有用但受控的方式扩展或覆盖继承的属性。
1. 类可以从 trait 继承。可以将 trait 视为轻量级类，允许以特定、有限的方式从多个超类继承。
1. （单例）对象是 Scala 类的一种特殊类型。
  1. 它们不是上面提到的对象。请记住，我们称那些为实例。

我们即将了解如何在 Scala 中创建类。

# 一个类示例
创建一个 Scala类的示例如下：

````scala
// WrapCounter 根据位数大小计数到最大值
class WrapCounter(counterBits: Int) {

  val max: Long = (1 << counterBits) - 1
  var counter = 0L
    
  def inc(): Long = {
    counter = counter + 1
    if (counter > max) {
        counter = 0
    }
    counter
  }
  println(s"计数器已创建，最大值为 $max")
}
````

这里有什么：
* ```class WrapCounter``` -- 这是 **WrapCounter** 的定义。
* ```(counterBits: Int)``` -- 创建一个 WrapCounter 需要一个整数参数，其名称很好地表明它是计数器的位宽。
* 大括号 ({}) 界定一个代码块。大多数类使用代码块来定义变量、常量和方法（函数）。
* `val max: Long =` -- 该类包含一个成员变量 **max**，声明为 `Long` 类型，并在创建类时初始化。
* `(1 << counterBits) - 1` 计算 **counterBits** 位可以包含的最大值。由于 **max** 是用 `val` 创建的，因此无法更改。
* 创建了一个变量 **counter** 并将其初始化为 **0L**。**L** 表示 0 是一个长整型值；因此，**counter** 被推断为 Long 类型。
* **max** 和 **counter** 通常称为类的_成员变量_。
* 定义了一个类方法 **inc**，它不接受任何参数并返回一个 **Long** 值。
* 方法 **inc** 的主体是一个代码块，其中包含：
  * `counter = counter + 1` 使 **counter** 递增。
  * `if (counter > max) { counter = 0 }` 测试计数器是否大于 **max** 值，如果是，则将其设置回零。
  * `counter` -- 代码块的最后一行很重要。
    * 作为代码块最后一行的任何表达式都被视为该代码块的返回值。调用语句可以使用或忽略该返回值。
    * 这非常普遍适用；例如，由于 `if` then `else` 语句使用代码块定义其 true 和 false 子句，因此它可以返回值，即 `val result = if (10 * 10 > 90) "greater" else "lesser"` 将创建一个值为“greater”的 `val`。
  * 因此，在这种情况下，函数 **inc** 返回 **counter** 的值。
* `println(s"counter created with max value $max")` 将一个字符串打印到标准输出。因为 **println** 直接在定义代码块中，所以它是类初始化代码的一部分，并且在每次创建此类的实例时都会运行，即打印出字符串。
* 在这种情况下打印的字符串是一个_内插_字符串。
  * 第一个双引号前面的前导 **s** 将此标识为内插字符串。
  * 内插字符串在运行时进行处理。
  * **\$max** 被替换为 max 的值。
  * 如果 **\$** 后跟一个代码块，则该代码块中可以包含任意 Scala 代码。
    * 例如，**`println(s"doubled max is ${max + max}")`**。
    * 此代码块的返回值将插入到 `${...}` 的位置。
    * 如果返回值不是字符串，它将被转换为字符串；Scala 中几乎每个类或类型都有到字符串的隐式转换定义）。
  * 通常应避免在每次创建类的实例时都打印某些内容，以避免淹没标准输出，除非您正在调试。

# 代码块
代码块由大括号界定。一个块可以包含零行或多行 Scala 代码。Scala 代码的最后一行成为代码块的返回值（可以忽略）。没有行的代码块将返回一个特殊的类似 null 的对象，称为 `Unit`。代码块在整个 Scala 中使用：它们是类定义的主体，它们构成函数和方法定义，它们是 `if` 语句的子句，并且它们是 `for` 和许多其他 Scala 运算符的主体。

## 参数化代码块
代码块可以接受参数。在类和方法定义的情况下，这些参数看起来像大多数传统编程语言中的参数。在下面的示例中，`c` 和 `s` 是代码块的参数。

````scala
// 单行代码块不需要用 {} 括起来
def add1(c: Int): Int = c + 1

class RepeatString(s: String) {
  val repeatedString = s + s
}
````

**重要提示**：代码块还有另一种参数化方式。这是一个示例。

````scala
val intList = List(1, 2, 3)
val stringList = intList.map { i =>
  i.toString
}
````

代码块被传递给 List 类的 `map` 方法。`map` 方法要求其代码块具有单个参数。为列表的每个成员调用代码块，并且代码块返回转换为字符串的该成员。Scala 对此语法的变体几乎过分接受。您可能会看到以许多不同方式编写此内容。这种类型的代码块称为匿名函数，有关匿名函数的更多详细信息将在后续模块中提供。

这里的目标是帮助您在遇到不同的符号类型时识别它们。随着您使用 Scala，这些符号会显得更加舒适和熟悉。作者倾向于偏爱特定的风格，并且在某些个别语法情况下，一种符号会显得更自然。单行代码倾向于使用更简洁的形式。复杂的代码块通常具有更具叙述性的外观。为了使协作更容易，应该考虑浏览 [Scala 风格指南](http://docs.scala-lang.org/style/) 中的最佳实践。

# 命名参数和参数默认值
考虑以下方法定义。
```scala
def myMethod(count: Int, wrap: Boolean, wrapValue: Int = 24): Unit = { ... }
```
调用该方法时，您通常会看到参数名称以及传入的值。
```scala
myMethod(count = 10, wrap = false, wrapValue = 23)
```
使用命名参数，您甚至可以以不同的顺序调用函数。
```scala
myMethod(wrapValue = 23, wrap = false, count = 10)
```
对于频繁调用的方法，参数顺序可能很明显。但对于不太常用的方法，尤其是布尔参数，在调用时包含名称可以使您的代码更具可读性。如果方法具有一长串相同类型的参数，使用名称也可以减少出错的机会。类定义的参数也使用这种命名参数方案（它们实际上只是类构造函数的参数）。

当某些参数具有默认值（不需要覆盖）时，调用者只需（按名称）传递不使用默认值的特定参数。请注意，参数 `wrapValue` 的默认值为 24。因此，

```scala
myMethod(wrap = false, count = 10)
```
将像传入 24 一样工作。

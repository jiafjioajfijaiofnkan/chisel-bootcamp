# 生成器和集合<a name="generators-and-collections"></a> 
在本节中，我们将重点关注*生成器*的概念以及使用 Scala 集合作为实现它们的工具。
我们将不再将 Chisel 代码视为电路的*实例*，即特定电路的描述，
而是将其视为电路的生成器。

我们将从先前练习中的 FIR 滤波器开始。

```scala
class My4ElementFir(b0: Int, b1: Int, b2: Int, b3: Int) extends Module {
  val io = IO(new Bundle {
    val in = Input(UInt(8.W))
    val out = Output(UInt(8.W))
  })

  val x_n1 = RegNext(io.in, 0.U)
  val x_n2 = RegNext(x_n1, 0.U)
  val x_n3 = RegNext(x_n2, 0.U)
  io.out := io.in * b0.U(8.W) + x_n1 * b1.U(8.W) +
    x_n2 * b2.U(8.W) + x_n3 * b3.U(8.W)
}
```

这个电路是生成器的一个简单例子，因为它可以生成具有
不同系数的这个 4 抽头滤波器的版本。但是如果我们想让电路有更多的抽头呢？我们将分几个步骤来完成这个。

- 构建一个抽头可配置 FIR 的软件*黄金模型*。
- 重新设计我们的测试以使用此模型，并确认其有效。
- 重构我们的 My4ElementFir 以允许可配置数量的抽头。
- 使用我们新的测试工具测试新电路。

<span style="color:blue">**示例：FIR 黄金模型**</span><br><a name="fir-golden-model"></a> 
下面是一个 FIR 电路的 Scala 软件实现。

```scala
/**
  * 一个具有任意数量抽头的 FIR 滤波器的简单实现。
  */
class ScalaFirFilter(taps: Seq[Int]) {
  var pseudoRegisters = List.fill(taps.length)(0)

  def poke(value: Int): Int = {
    pseudoRegisters = value :: pseudoRegisters.take(taps.length - 1)
    var accumulator = 0
    for(i <- taps.indices) {
      accumulator += taps(i) * pseudoRegisters(i)
    }
    accumulator
  }
}
```

### Seq
请注意，`taps` 已成为 `Seq[Int]`，这意味着类的用户在构造类时可以传递任意长度的 `Int` 序列。
### 寄存器
通过 `  var pseudoRegisters = List.fill(taps.length)(0)`，我们创建了一个 `List`，它将保存先前周期的值。选择 `List` 是因为其在头部添加元素和移除最后一个元素的语法非常简单。几乎可以使用 scala 集合家族的任何成员。我们还将此列表初始化为包含所有零。
### Poke
我们的类添加了一个 poke 函数/方法，用于模拟将新输入放入滤波器并使时钟循环。
### 更新寄存器
行 `pseudoRegisters = value :: pseudoRegisters.take(taps.length - 1)` 首先使用列表的 `take` 方法保留列表除最后一个元素之外的所有元素，然后使用 `::` 列表连接运算符将 `value` 添加到列表的缩减版本的头部。
### 计算输出
一个简单的 for 循环和一个累加器对列表的每个元素乘以其对应的抽头系数求和。仅包含 `accumulator` 的行将该值作为函数结果返回。
## 修改我们之前的测试以测试我们的黄金模型
现在我们将使用我们之前的工作来确认我们的黄金模型有效。一些编辑魔法将我们之前的测试工具转换为...

```scala
val filter = new ScalaFirFilter(Seq(1, 1, 1, 1))

var out = 0

out = filter.poke(1)
println(s"out = $out")
assert(out == 1)  // 1, 0, 0, 0

out = filter.poke(4)
assert(out == 5)  // 4, 1, 0, 0
println(s"out = $out")

out = filter.poke(3)
assert(out == 8)  // 3, 4, 1, 0
println(s"out = $out")

out = filter.poke(2)
assert(out == 10)  // 2, 3, 4, 1
println(s"out = $out")

out = filter.poke(7)
assert(out == 16)  // 7, 2, 3, 4
println(s"out = $out")

out = filter.poke(0)
assert(out == 12)  // 0, 7, 2, 3
println(s"out = $out")
```

执行前一个块表明我们的软件模型返回与 My4ElementFir 相同的结果。

## 使用黄金模型测试电路。<a name="use-golden-model-as-test"></a> 
既然我们对我们的黄金模型相当有信心，我们将重写我们的测试，以将电路输出与黄金模型的输出进行比较，而不是使用费力手工制作的示例。
以下是快速的第一次尝试。

```scala
val goldenModel = new ScalaFirFilter(Seq(1, 1, 1, 1))

test(new My4ElementFir(1, 1, 1, 1)) { c =>
    for(i <- 0 until 100) {
        val input = scala.util.Random.nextInt(8)

        val goldenModelResult = goldenModel.poke(input)

        c.io.in.poke(input.U)

        c.io.out.expect(goldenModelResult.U, s"i $i, input $input, gm $goldenModelResult, ${c.io.out.peek().litValue}")

        c.clock.step(1)
    }

}
```

我们的测试运行 100 个周期，并检查硬件和软件这两种不同方法在每个步骤中是否同步。

### 需要注意的事项
（即我们在编写此内容时实际犯的错误。）

1. 将步骤放在正确的位置。软件和硬件的执行方式不同；很容易弄错。
2. 此测试很弱，因为它对 IO 和寄存器的大小非常敏感。实现一个在任意数据位宽下观察环绕行为的软件黄金模型可能很复杂。这里我们只确保只传入适合的值。

<span style="color:blue">**示例：参数化 FIR 生成器**</span><br><a name="fir-golden-model"></a> 
下面我们创建了一个新的 Filter 类 `MyManyElementsFilter`，它接受一个 `Seq` 常量作为抽头。这个列表可以是任意数量的元素。
为了方便起见，添加了一个 `bitWidth`，它允许我们控制电路可以处理的数字大小。
为了响应可变长度，我们不得不重构寄存器的创建方式以及它们的连接方式。
下面使用的方法论使用了可用集合函数库的一个简单子集。
后续章节将展示如何更简洁地表达行为，同时也能更清楚地说明正在发生的事情。

```scala
class MyManyElementFir(consts: Seq[Int], bitWidth: Int) extends Module {
  val io = IO(new Bundle {
    val in = Input(UInt(bitWidth.W))
    val out = Output(UInt(bitWidth.W))
  })

  val regs = mutable.ArrayBuffer[UInt]()
  for(i <- 0 until consts.length) {
      if(i == 0) regs += io.in
      else       regs += RegNext(regs(i - 1), 0.U)
  }
  
  val muls = mutable.ArrayBuffer[UInt]()
  for(i <- 0 until consts.length) {
      muls += regs(i) * consts(i).U
  }

  val scan = mutable.ArrayBuffer[UInt]()
  for(i <- 0 until consts.length) {
      if(i == 0) scan += muls(i)
      else scan += muls(i) + scan(i - 1)
  }

  io.out := scan.last
}
```

#### 我们是如何做到的
从第 7、13 和 18 行开始有三个平行的部分。
我们使用的是一种名为 `ArrayBuffer` 的 Scala 集合类型。
`ArrayBuffer` 允许您使用 `+=` 运算符追加元素（也可以插入和删除，但我们不需要这些）。
首先，我们创建一个 ArrayBuffer `regs`，其元素将是 `UInt`。
然后遍历抽头，将输入作为第一个元素添加，然后使用 RegNext 创建寄存器，该寄存器将寄存器的输入连接到前一个元素 (`regs(i-1)`) 并将其初始化为无符号零 (`0.U`)。
这些寄存器将根据需要保存输入的先前值。

接下来，我们创建另一个 `UInt` 的 ArrayBuffer `muls`。
muls 的每个元素将是一个节点，其第 i 个元素是 `regs(i)` 和 `const(i)` 的乘积。

注意 `scan.last` 方法的使用。
它获取集合的最后一个元素，并且是在 `regs` 构建期间使用的 `regs(i - 1)` 的更优雅的替代方法。

### 它的行为是否与 `My4ElementFir` 相同？
对我们新版本的一个很好的初步测试是看看它是否能通过我们刚刚应用于
`My4ElementFir` 的测试。
我们创建一个 `MyManyElementFir` 的实例，并通过它运行更多的数据。

```scala
val goldenModel = new ScalaFirFilter(Seq(1, 1, 1, 1))

test(new MyManyElementFir(Seq(1, 1, 1, 1), 8)) { c =>
    for(i <- 0 until 100) {
      val input = scala.util.Random.nextInt(8)

      val goldenModelResult = goldenModel.poke(input)

      c.io.in.poke(input.U)

      c.io.out.expect(goldenModelResult.U, s"i $i, input $input, gm $goldenModelResult, ${c.io.out.peek().litValue}")

      c.clock.step(1)
    }
}
```

### 现在让我们测试一堆不同大小的 FIR 滤波器
我们创建一些辅助函数：`r` 用于获取随机数；`runOneTest` 用于为特定抽头集创建黄金模型和滤波器的硬件仿真，然后通过滤波器运行至少两倍于抽头数量的数据。

```scala
/** 获取随机整数的便捷方法
  */
def r(): Int = {
  scala.util.Random.nextInt(1024)
}

/**
  * 运行比较软件和硬件滤波器的测试
  * 运行至少两倍于抽头数量的样本
  */
def runOneTest(taps: Seq[Int]) {
    val goldenModel = new ScalaFirFilter(taps)

    test(new MyManyElementFir(taps, 32)) { c =>
        for(i <- 0 until 2 * taps.length) {
            val input = r()

            val goldenModelResult = goldenModel.poke(input)

            c.io.in.poke(input.U)

            c.io.out.expect(goldenModelResult.U, s"i $i, input $input, gm $goldenModelResult, ${c.io.out.peek().litValue}")

            c.clock.step(1)
        }
    }
}

for(tapSize <- 2 until 100 by 10) {
    val taps = Seq.fill(tapSize)(r())  // 创建一个随机系数序列

    runOneTest(taps)
}
```

# 硬件集合

<span style="color:blue">**示例：向我们的 FIR 添加运行时可配置的抽头**</span><br>
以下代码向我们的 FIR 生成器的 IO 添加了一个额外的 `consts` 向量，该向量允许在电路生成后从外部更改系数。
这是通过 Chisel 集合类型 `Vec` 完成的。
`Vec` 支持许多 scala 集合方法，但它只能包含 Chisel 硬件元素。
`Vec` 仅应在普通 Scala 集合不起作用的情况下使用。
基本上，这适用于以下两种情况之一：
1. 您需要在 Bundle 中包含元素集合，通常是作为 IO 使用的 Bundle。
2. 您需要通过作为硬件一部分的索引访问集合（例如寄存器文件）。


```scala
class MyManyDynamicElementVecFir(length: Int) extends Module {
  val io = IO(new Bundle {
    val in = Input(UInt(8.W))
    val out = Output(UInt(8.W))
    val consts = Input(Vec(length, UInt(8.W)))
  })

  // 参考解决方案
  val regs = RegInit(VecInit(Seq.fill(length - 1)(0.U(8.W))))
  for(i <- 0 until length - 1) {
      if(i == 0) regs(i) := io.in
      else       regs(i) := regs(i - 1)
  }
  
  val muls = Wire(Vec(length, UInt(8.W)))
  for(i <- 0 until length) {
      if(i == 0) muls(i) := io.in * io.consts(i)
      else       muls(i) := regs(i - 1) * io.consts(i)
  }

  val scan = Wire(Vec(length, UInt(8.W)))
  for(i <- 0 until length) {
      if(i == 0) scan(i) := muls(i)
      else scan(i) := muls(i) + scan(i - 1)
  }

  io.out := scan(length - 1)
}
```

```scala
val goldenModel = new ScalaFirFilter(Seq(1, 1, 1, 1))

test(new MyManyDynamicElementVecFir(4)) { c =>
    c.io.consts(0).poke(1.U)
    c.io.consts(1).poke(1.U)
    c.io.consts(2).poke(1.U)
    c.io.consts(3).poke(1.U)
    for(i <- 0 until 100) {
        val input = scala.util.Random.nextInt(8)

        val goldenModelResult = goldenModel.poke(input)

        c.io.in.poke(input.U)

        c.io.out.expect(goldenModelResult.U, s"i $i, input $input, gm $goldenModelResult, ${c.io.out.peek().litValue}")

        c.clock.step(1)
    }
}
```

[寄存器文件](https://en.wikipedia.org/wiki/Register_file) 是构建处理器的重要组成部分。
寄存器文件是一个寄存器数组，可以通过多个读或写端口进行读写。
每个端口都包含一个地址和数据字段。

[RISC-V 指令集架构](https://riscv.org/specifications/) 定义了几个变体，其中最简单的是 RV32I。
RV32I 有一个包含 32 个 32 位寄存器的数组。
**索引为 0 的寄存器（第一个寄存器）在读取时始终为零，无论您向其写入什么内容**（通常 иметь 0 под рукой полезно）。

# Scala 中的函数式编程
Scala 函数在模块 1 中已介绍，并且您在上一模块中也看到了它们的广泛使用。这里是对函数的复习。函数接受任意数量的输入并产生一个输出。输入通常称为函数的参数。要不产生输出，请返回 `Unit` 类型。

<span style="color:blue">**示例：自定义函数**</span><br>
以下是 Scala 中的一些函数示例。

```scala
// 无输入或输出（两个版本）。
def hello1(): Unit = print("Hello!")
def hello2 = print("Hello again!")

// 数学运算：一个输入和一个输出。
def times2(x: Int): Int = 2 * x

// 输入可以有默认值，显式指定返回类型是可选的。
// 注意，我们建议指定返回类型以避免意外/错误。
def timesN(x: Int, n: Int = 2) = n * x

// 调用上面列出的函数。
hello1()
hello2
times2(4)
timesN(4)         // 无需指定 n 即可使用默认值
timesN(4, 3)      // 参数顺序与函数定义时的顺序相同
timesN(n=7, x=2)  // 参数可以重新排序并显式赋值
```

## 函数作为对象
Scala 中的函数是一等对象。这意味着我们可以将函数赋给 `val`，并将其作为参数传递给类、对象或其他函数。

<span style="color:blue">**示例：函数对象**</span><br>
以下是作为函数和作为对象实现的相同函数。

```scala
// 这些是普通函数。
def plus1funct(x: Int): Int = x + 1
def times2funct(x: Int): Int = x * 2

// 这些是作为 val 的函数。
// 第一个显式指定返回类型。
val plus1val: Int => Int = x => x + 1
val times2val = (x: Int) => x * 2

// 两者的调用方式相同。
plus1funct(4)
plus1val(4)
plus1funct(x=4)
//plus1val(x=4) // 这不起作用
```

为什么要创建一个 `val` 而不是 `def`？使用 `val`，您现在可以将函数传递给其他函数，如下所示。您甚至可以创建自己的接受其他函数作为参数的函数。形式上，接受或产生函数的函数称为*高阶函数*。您在上一模块中已经看到它们的使用，但现在您将创建自己的高阶函数！

<span style="color:blue">**示例：高阶函数**</span><br>
这里我们再次展示 `map`，并且还创建了一个新函数 `opN`，它接受一个函数 `op` 作为参数。

```scala
// 创建我们的函数
val plus1 = (x: Int) => x + 1
val times2 = (x: Int) => x * 2

// 将其传递给 map，一个列表函数
val myList = List(1, 2, 5, 9)
val myListPlus = myList.map(plus1)
val myListTimes = myList.map(times2)

// 创建一个自定义函数，使用递归对 X 执行 N 次操作
def opN(x: Int, n: Int, op: Int => Int): Int = {
  if (n <= 0) { x }
  else { opN(op(x), n-1, op) }
}

opN(7, 3, plus1)
opN(7, 3, times2)
```

<span style="color:blue">**示例：函数与对象**</span><br>
使用不带参数的函数时，可能会出现一种令人困惑的情况。函数在每次调用时都会被求值，而 `val` 则在实例化时被求值。

```scala
import scala.util.Random

// x 和 y 都调用 nextInt 函数，但 x 立即求值，而 y 是一个函数
val x = Random.nextInt
def y = Random.nextInt

// x 之前已求值，因此它是一个常量
println(s"x = $x")
println(s"x = $x")

// y 是一个函数，每次调用都会重新求值，因此这些会产生不同的结果
println(s"y = $y")
println(s"y = $y")
```

```scala
val myList = List(5, 6, 7, 8)

// 使用匿名函数将列表中的每个项目加一
// 参数传递给下划线变量
// 这些都做同样的事情
myList.map( (x:Int) => x + 1 )
myList.map(_ + 1)

// 一种常见情况是在匿名函数中使用 case 语句
val myAnyList = List(1, 2, "3", 4L, myList)
myAnyList.map {
  case (_:Int|_:Long) => "数字"
  case _:String => "字符串"
  case _ => "错误"
}
```

# Chisel 中的函数式编程
让我们看一些在 Chisel 中创建硬件生成器时如何使用函数式编程的示例。

<span style="color:blue">**示例：FIR 滤波器**</span><br>
首先，我们将回顾前面示例中的 FIR 滤波器。我们将向 FIR 传递一个定义如何计算窗口系数的函数，而不是将系数作为参数传递给类或使其可编程。此函数将采用窗口长度和位宽来生成缩放后的系数列表。以下是两个示例窗口。为避免分数，我们将系数缩放为介于最大和最小整数值之间。有关这些窗口的更多信息，请查看[此维基百科页面](https://en.wikipedia.org/wiki/Window_function)。

```scala
// 获取一些数学函数
import scala.math.{abs, round, cos, Pi, pow}

// 简单的三角窗
val TriangularWindow: (Int, Int) => Seq[Int] = (length, bitwidth) => {
  val raw_coeffs = (0 until length).map( (x:Int) => 1-abs((x.toDouble-(length-1)/2.0)/((length-1)/2.0)) )
  val scaled_coeffs = raw_coeffs.map( (x: Double) => round(x * pow(2, bitwidth)).toInt)
  scaled_coeffs
}

// 汉明窗
val HammingWindow: (Int, Int) => Seq[Int] = (length, bitwidth) => {
  val raw_coeffs = (0 until length).map( (x: Int) => 0.54 - 0.46*cos(2*Pi*x/(length-1)))
  val scaled_coeffs = raw_coeffs.map( (x: Double) => round(x * pow(2, bitwidth)).toInt)
  scaled_coeffs
}

// 看看！第一个参数是窗口长度，第二个参数是位宽
TriangularWindow(10, 16)
HammingWindow(10, 16)
```

现在我们将创建一个 FIR 滤波器，它接受一个窗口函数作为参数。这使我们以后可以定义新的窗口并保留相同的 FIR 生成器。它还允许我们独立调整 FIR 的大小，因为知道窗口会针对不同的长度或位宽重新计算。由于我们在编译时选择窗口，因此这些系数是固定的。

```scala
// 我们的 FIR 具有参数化的窗口长度、IO 位宽和加窗函数
class MyFir(length: Int, bitwidth: Int, window: (Int, Int) => Seq[Int]) extends Module {
  val io = IO(new Bundle {
    val in = Input(UInt(bitwidth.W))
    val out = Output(UInt((bitwidth*2+length-1).W)) // 预期位增长，保守但懒惰
  })

  // 使用提供的窗口函数计算系数，映射到 UInt
  val coeffs = window(length, bitwidth).map(_.U)
  
  // 创建一个数组来保存延迟的输出
  // 注意：我们在这里避免使用 Vec，因为我们不需要动态索引
  val delays = Seq.fill(length)(Wire(UInt(bitwidth.W))).scan(io.in)( (prev: UInt, next: UInt) => {
    next := RegNext(prev)
    next
  })
  
  // 相乘，将结果放入“mults”
  val mults = delays.zip(coeffs).map{ case(delay: UInt, coeff: UInt) => delay * coeff }
  
  // 将乘法器输出与位增长相加
  val result = mults.reduce(_+&_)

  // 连接输出
  io.out := result
}

visualize(() => new MyFir(7, 12, TriangularWindow))
```

最后三行可以很容易地合并为一行。还要注意我们如何保守地处理位宽增长以避免丢失。

<span style="color:blue">**示例：FIR 滤波器测试器**</span><br>
让我们测试一下我们的 FIR！以前，我们提供了一个自定义的黄金模型。这次我们将使用 Breeze（一个包含有用的线性代数和信号处理函数的 Scala 库）作为我们 FIR 滤波器的黄金模型。下面的代码将 Chisel 输出与黄金模型输出进行比较，任何错误都会导致测试器失败。

尝试取消注释 expect 调用之后末尾的 print 语句。还要尝试将窗口从三角窗更改为汉明窗。

```scala
// 数学导入
import scala.math.{pow, sin, Pi}
import breeze.signal.{filter, OptOverhang}
import breeze.signal.support.{CanFilter, FIRKernel1D}
import breeze.linalg.DenseVector

// 测试参数
val length = 7
val bitwidth = 12 // 必须小于 15，否则 Int 无法表示数据，需要 BigInt
val window = TriangularWindow

// 测试我们的 FIR
test(new MyFir(length, bitwidth, window)) { c =>
    
    // 测试数据
    val n = 100 // 输入长度
    val sine_freq = 10
    val samp_freq = 100

    // 采样数据，缩放到 0 和 2^bitwidth 之间
    val max_value = pow(2, bitwidth)-1
    val sine = (0 until n).map(i => (max_value/2 + max_value/2*sin(2*Pi*sine_freq/samp_freq*i)).toInt)
    //println(s"input = ${sine.toArray.deep.mkString(", ")}")

    // 系数
    val coeffs = window(length, bitwidth)
    //println(s"coeffs = ${coeffs.toArray.deep.mkString(", ")}")

    // 使用 breeze 滤波器作为黄金模型；需要反转系数
    val expected = filter(
        DenseVector(sine.toArray),
        FIRKernel1D(DenseVector(coeffs.reverse.toArray), 1.0, ""),
        OptOverhang.None
    )
    expected.toArray // 似乎是必需的
    //println(s"exp_out = ${expected.toArray.deep.mkString(", ")}") // 这似乎是必需的

    // 将数据推入我们的 FIR 并检查结果
    c.reset.poke(true.B)
    c.clock.step(5)
    c.reset.poke(false.B)
    for (i <- 0 until n) {
        c.io.in.poke(sine(i).U)
        if (i >= length-1) { // 等待所有寄存器初始化，因为我们没有对数据进行零填充
            val expectValue = expected(i-length+1)
            //println(s"expected value is $expectValue")
            c.io.out.expect(expected(i-length+1).U)
            //println(s"cycle $i, got ${c.io.out.peek()}, expect ${expected(i-length+1)}")
        }
        c.clock.step(1)
    }
}
```
